```python
import pandas as pd
import numpy as np
import os
```

```python
DATA_PATH = "../data/raw"

employee_df = pd.read_csv(f"{DATA_PATH}/employee_attrition.csv")
engagement_df = pd.read_csv(f"{DATA_PATH}/hr_performance_engagement.csv")
occupation_df = pd.read_csv(f"{DATA_PATH}/occupation_data.csv")
essential_skills_df = pd.read_csv(f"{DATA_PATH}/essential_skills.csv")
software_skills_df = pd.read_csv(f"{DATA_PATH}/software_skills.csv")
```

```python
# SCHEMA CHECK

expected_columns = [
    "EmployeeID",
    "Name",
    "Gender",
    "Age",
    "Department",
    "JobRole",
    "EducationLevel",
    "JoiningDate",
    "CountryCode",
    "Country",
    "PhoneNumber",
    "MonthlySalary",
    "OvertimeHoursPerMonth",
    "LeavesTaken",
    "LastLeaveDate",
    "LeaveDayName",
    "ProjectsHandled",
    "TrainingHours",
    "CustomerSatisfaction",
    "LastPromotionYear",
    "YearsAtCompany",
    "WorkLifeBalanceScore",
    "PerformanceRating",
    "AttritionRisk"
]

missing_columns = set(expected_columns) - set(employee_df.columns)

assert not missing_columns, f"Missing expected columns: {missing_columns}"
```

```python
# TYPE CHECK

assert pd.api.types.is_integer_dtype(
    employee_df["Age"]
), "Age must be an integer"

assert pd.api.types.is_numeric_dtype(
    employee_df["MonthlySalary"]
), "MonthlySalary must be numeric"

assert pd.api.types.is_string_dtype(
    employee_df["Department"]
), "Department must be text"
```

```python
# RANGE CHECK
assert employee_df["Age"].between(18, 100).all(), "Age out of range"
```

```python
# UNIQUENESS
assert employee_df["EmployeeID"].is_unique, "Duplicate EmployeeID found"
```

```python
# CATEGORY CHECK
assert set(employee_df["AttritionRisk"].dropna().unique()) <= {"Yes", "No"}, \
    "Unexpected AttritionRisk value"
```

### Engagement Data Validation

```python
# SHAPE CHECK
assert engagement_df.shape[0] > 0, "Engagement data is empty"
assert engagement_df.shape[1] > 0, "Engagement data has no columns"
```

```python
# NO NULLS IN KEY
assert engagement_df["Employee ID"].notna().all(), "Employee ID has nulls in engagement data"
```

### Occupation Data Validation

```python
# SHAPE CHECK
assert occupation_df.shape[0] > 0, "Occupation data is empty"
assert occupation_df.shape[1] > 0, "Occupation data has no columns"
```

```python
# UNIQUENESS
assert occupation_df["O*NET-SOC Code"].is_unique, "Duplicate O*NET-SOC Code found in occupation data"
```

### Essential Skills Data Validation

```python
# SHAPE CHECK
assert essential_skills_df.shape[0] > 0, "Essential skills data is empty"
assert essential_skills_df.shape[1] > 0, "Essential skills data has no columns"
```

```python
# NO NULLS IN KEY
assert essential_skills_df["O*NET-SOC Code"].notna().all(), "O*NET-SOC Code has nulls in essential skills data"
```
