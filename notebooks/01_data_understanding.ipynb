```python
# EmployeeID overlap check between employee_attrition and engagement tables
employee_ids = set(employee_df["EmployeeID"])
engagement_ids = set(engagement_df["Employee ID"])
overlap = employee_ids & engagement_ids
print(f"Employee IDs in employee_attrition: {len(employee_ids)} (range {employee_df['EmployeeID'].min()}-{employee_df['EmployeeID'].max()})")
print(f"Employee IDs in engagement: {len(engagement_ids)} (range {engagement_df['Employee ID'].min()}-{engagement_df['Employee ID'].max()})")
print(f"Overlap: {len(overlap)}")
if len(overlap) == 0:
    print("WARNING: These two tables use different ID spaces and cannot be joined on EmployeeID.")
```

# End-of-Day Data Understanding Report

## 1. employee_attrition.csv

- **Row/column count:** 500 rows × 24 columns
- **Likely primary key:** `EmployeeID`
- **Missing values:** `CustomerSatisfaction` has 319 missing values. All other columns have 0 missing values.
- **Purpose:** Feeds the attrition ML model directly.
- **Potential join:** `EmployeeID` may connect to `hr_performance_engagement.csv` through `Employee ID`.

## 2. hr_performance_engagement.csv

- **Row/column count:** 3,150 rows × 39 columns
- **Likely primary key:** `Employee ID`
- **Missing values:** `ExitDate` has 1,544 missing values and `TerminationDescription` has 1,544 missing values. All other columns have 0 missing values.
- **Purpose:** Engagement analytics; no ML needed initially.
- **Potential join:** `Employee ID` may connect to `employee_attrition.csv` through `EmployeeID`.

## 3. occupation_data.csv

- **Row/column count:** 1,016 rows × 3 columns
- **Likely primary key:** No obvious ID column was identified.
- **Missing values:** No missing values.
- **Purpose:** Becomes the role master reference table.
- **Potential join:** `O*NET-SOC Code` may connect this dataset to the skills datasets.

## 4. essential_skills.csv

- **Row/column count:** 18,200 rows × 15 columns
- **Likely primary key:** No single primary key was established from the ID-column check.
- **Missing values:** No missing values.
- **Purpose:** Provides essential-skills information for the required-skills data per role.
- **Potential join:** `O*NET-SOC Code` may connect this dataset to `occupation_data.csv`.

## 5. software_skills.csv

- **Row/column count:** 31,821 rows × 7 columns
- **Likely primary key:** No single primary key was established from the ID-column check.
- **Missing values:** No missing values.
- **Purpose:** Provides software-skills information for the required-skills data per role.
- **Potential join:** `O*NET-SOC Code` may connect this dataset to `occupation_data.csv`.

## Overall Data Understanding Conclusion

All five datasets have been loaded and inspected for shape, columns, data types, missing values, duplicate rows, and candidate ID columns.

The main candidate employee relationship is:

`employee_attrition.csv.EmployeeID` ↔ `hr_performance_engagement.csv.Employee ID`
